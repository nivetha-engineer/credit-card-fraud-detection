```python
x=df_copy.drop('Class',axis=1)
y=df_copy['Class']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
x_train=scalar.fit_transform(x_train)
x_test=scalar.transform(x_test)
```

```python
from imblearn.over_sampling import SMOTE
oversampling=SMOTE(random_state=42)
x_train,y_train=oversampling.fit_resample(x_train,y_train)
```

```python
y_train.value_counts()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,roc_auc_score)
```

```python
model={
    "LogisticRegression":LogisticRegression( random_state=42,solver="liblinear",max_iter=1000),
    "DecisionTreeClassifier":DecisionTreeClassifier(random_state=42),
    "RandomForestClassifier":RandomForestClassifier(n_estimators=100,min_samples_split=20,max_features=5,max_depth=10,random_state=42),
    "XGBoost": XGBClassifier(n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    eval_metric="logloss")
        
}
```

```python
result=[]
for name,models in model.items():
    print("="*45)
    print(f"Training {models}")
    models.fit(x_train,y_train)
    print(f"{name} Finished")
    y_pred=models.predict(x_test)
    y_prob=models.predict_proba(x_test)[:,1]
    result.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })
print("finished")
    
```
